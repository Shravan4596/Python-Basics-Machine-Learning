# 1️⃣ Creating Functions

## 🎯 Learning Objective

Learn how to define and call functions.

### Syntax

```python
def function_name():
    # code
```

````python
# Function without parameters

def welcome():
    print("Welcome to Python Functions!")

welcome()
````

````python
# Function to display course information

def course():
    print("Course: Python Programming")
    print("Duration: 3 Months")

course()
````

# 2️⃣ Function Parameters

## 🎯 Learning Objective

Pass values into functions using parameters.

Parameters make functions flexible and reusable.

````python
def greet(name):
    print("Hello,", name)

greet("Shravan")
````

````python
def addition(a, b):
    print("Sum =", a + b)

addition(20, 30)
````

# Example 1: Positional Arguments

````python
def student(name, age):
    print("Name :", name)
    print("Age  :", age)

student("Shravan", 24)
````

# Example 2: Keyword Arguments

````python
def employee(name, salary):
    print("Employee :", name)
    print("Salary   :", salary)

employee(salary=45000, name="Rahul")
````

# Example 3: Default Arguments

````python
def country(name, country="India"):
    print(name, "belongs to", country)

country("Shravan")
country("John", "USA")
````

 # Example 4: Variable-Length Arguments

````python
def total_marks(*marks):
    print("Total =", sum(marks))

total_marks(80, 90, 95, 88)
````

#  Return Statement

## 🎯 Learning Objective

Learn how to return values from a function.

The `return` statement sends a value back to the caller.

````python
def square(number):
    return number ** 2

result = square(5)

print(result)
````

````python
def largest(a, b):
    if a > b:
        return a
    else:
        return b

print("Largest =", largest(25, 40))
````

#  Lambda Functions

## 🎯 Learning Objective

Learn how to create anonymous (lambda) functions.

Lambda functions are useful for short, one-line operations.

### Syntax

```python
lambda arguments : expression
```

````python
# Example 1
square = lambda x: x ** 2

print(square(8))
````

````python
# Example 2
addition = lambda a, b: a + b

print(addition(15, 35))
````

# Recursion

## 🎯 Learning Objective

Understand how a function can call itself.

A recursive function must have a stopping condition (base case).

````python
def countdown(n):

    if n == 0:
        print("Done")
        return

    print(n)

    countdown(n - 1)

countdown(5)
````

````python
def factorial(n):

    if n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
````

#  Real-world Examples

Functions are widely used in software development. Below are some practical examples.

````python
def calculator(a, b):
    print("Addition       :", a + b)
    print("Subtraction    :", a - b)
    print("Multiplication :", a * b)
    print("Division       :", a / b)

calculator(20, 5)
````

````python
def grade(marks):

    if marks >= 90:
        return "A"

    elif marks >= 80:
        return "B"

    elif marks >= 70:
        return "C"

    else:
        return "Fail"

print("Grade :", grade(87))
````

````python
# Example : Employee Bonus

def calculate_bonus(salary):

    bonus = salary * 0.10

    return bonus

print("Bonus =", calculate_bonus(50000))
````

````python
# Example : Area of Circle
def area_of_circle(radius):

    pi = 3.14159

    return pi * radius ** 2

print("Area =", round(area_of_circle(5), 2))
````

````python
# Example : Check Even or Odd

def check_even(number):

    if number % 2 == 0:
        return "Even"

    return "Odd"

print(check_even(25))
````
