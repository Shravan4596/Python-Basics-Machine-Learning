```python
#Program to find the sum of First N Natural Numbers
#Method 1:

num=int(input("Enter the value:"))
value=0
for i in range(1,num+1):
    value=value+i
    print("Sum of N natural numbers:",value)
```

```python
#Program to check Leap Year or Not

year=int(input("Enter Year:"))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print("yes,{} is leap year".format(year))
        else:
            print("no,{} is leap year".format(year))
    else:
            print("no,{} is leap year".format(year))
else:
            print("no,{} is leap year".format(year))
```

```python
#Program to find a Given Number is Prime or Not

num=int(input("Enter the number:"))
for i in range(2,num//2):
    if num%i==0:
        print("Not a Prime Number")
        break
else:
    print("It is a Prime Number")
```

```python
#Program to Print Prime Numbers in a Given Range

first=int(input("Enter the first Number:"))
second=int(input("Enter the second Number:"))
for i in range(first,second):
    for j in range(2, i//2):
        if i%j==0:
            break
else:
    print("prime number",i)
```

```python
#Program to find Sum of Digits Of a Number

num=[int(d) for d in input("enter the number:")]
sum=0
for i in range(0,len(num)):
    sum=sum+num[i]
    print("Sum of digits of a number:{}".format(sum))
```

```python
#Program to find the Reverse of a given number

num=int(input("Enter the number:"))
temp=num
reverse=0
while num>0:
    remainder=num%10
    reverse=(reverse*10)+remainder
    num=num//10
    print("The given number is {} and reverse is {}".format(temp,reverse))
```

```python
#Program to find a Number is Polindrome or not

number=int(input("enter the number:"))
temp=number
reverse=0
while number>0:
    remainder=number%10
    reverse=(reverse*10)+remainder
    number=number//10
if temp==reverse:
        print("given number {} is polindrome".format(temp))
else:
    print("given number {} is not a ppolindrome".format(temp))
```

```python
#programto find a Number is Armstrong or not

import math
value=int(input("Enter the Number:"))
num=[int(d) for d in str(value)]
sum=0
for i in range(0,len(num)):
    sum=sum+math.pow(num[i],len(num))
if sum==value:
        print("Given number is Armstrong Number")
else:
    print("Given Number is Not a Armstrong Number")
```
