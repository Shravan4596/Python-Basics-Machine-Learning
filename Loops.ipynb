## 📖 Introduction to Loops

A **loop** is a programming construct that allows you to execute a block of code repeatedly.

Loops help reduce code duplication and make programs more efficient.

Python supports two types of loops:

- **for loop** – Used when the number of iterations is known.
- **while loop** – Used when the number of iterations depends on a condition.

### 💡 Real-world Applications

- Reading records from a database
- Processing CSV files
- Sending emails to multiple users
- Machine Learning dataset iteration
- Data Analysis and Automation

# 1️⃣ `for` Loop

## 🎯 Learning Objective

Learn how to iterate over sequences such as lists, strings, tuples, and ranges.

---

## 📖 Theory

A `for` loop executes a block of code once for every item in a sequence.

### Syntax
```python
for variable in sequence:
    statements
```

````python
# Example 1: Print Numbers from 1 to 10

for number in range(1, 11):
    print(number)
````

````python
# Example 2: Iterate Through a List

fruits = ["Apple", "Banana", "Orange", "Mango"]

for fruit in fruits:
    print(fruit)
````

# 2️⃣ `while` Loop

## 🎯 Learning Objective

Learn how to execute a block of code repeatedly while a condition remains `True`.

---

## 📖 Theory

A `while` loop continues executing until its condition becomes `False`.

It is useful when the number of iterations is **not known in advance**.

### Syntax

```python
while condition:
    statements
```

````python
# Example 1: Print Numbers from 1 to 5

count = 1

while count <= 5:
    print(count)
    count += 1
````

````python
# Example 2: Countdown Timer

count = 5

while count > 0:
    print(count)
    count -= 1

print("Time's Up!")
````

# 3️⃣ Nested Loops

## 🎯 Learning Objective

Learn how to place one loop inside another loop.

---

## 📖 Theory

A nested loop is a loop inside another loop.

For every iteration of the outer loop, the inner loop executes completely.

### Syntax

```python
for i in range(...):
    for j in range(...):
        statements
```

````python
# Example 1: Multiplication Table

for i in range(1, 4):

    print(f"\nTable of {i}")

    for j in range(1, 11):

        print(f"{i} x {j} = {i*j}")
````

````python
# Example 2: Cartesian Coordinates

for x in range(3):
    for y in range(3):
        print(f"({x}, {y})")
````

# 4️⃣ break Statement

## 🎯 Learning Objective

Learn how to terminate a loop immediately using the `break` statement.

---

## 📖 Theory

The `break` statement is used to exit a loop before it completes all iterations.

Once `break` is executed, Python immediately exits the loop and continues with the next statement after the loop.

### Syntax

```python
for item in sequence:
    if condition:
        break
```

````python
# Example 1: Stop when number becomes 5

for number in range(1, 11):

    if number == 5:
        break

    print(number)
````

````python
# Example 2: Search for a Student

students = ["Rahul", "Sravan", "Anitha", "Kiran"]

for student in students:

    if student == "Sravan":
        print("Student Found")
        break
````

# 5️⃣ continue Statement

## 🎯 Learning Objective

Learn how to skip the current iteration and continue with the next iteration.

---

## 📖 Theory

The `continue` statement skips the current iteration without terminating the loop.

The loop proceeds to the next iteration.

### Syntax

```python
for item in sequence:

    if condition:
        continue
```

````python
# Example 1: Skip Number 5

for number in range(1,11):

    if number == 5:
        continue

    print(number)
````

````python
# Example 2: Print Only Odd Numbers

for number in range(1,11):

    if number % 2 == 0:
        continue

    print(number)
````

# 6️⃣ pass Statement

## 🎯 Learning Objective

Understand how to use the `pass` statement as a placeholder.

---

## 📖 Theory

The `pass` statement performs no action.

It is used when a statement is syntactically required but no code has been written yet.

````python
# Example 1: pass inside loop

for number in range(5):

    if number == 3:
        pass

    print(number)
````

````python
# Example 2: Empty if Block

number = 10

if number > 5:
    pass

print("Program Completed")
````

# 7️⃣ range() Function

## 🎯 Learning Objective

Learn how to generate sequences of numbers using the `range()` function.

---

## 📖 Theory

The `range()` function generates a sequence of integers.

### Syntax

```python
range(stop)

range(start, stop)

range(start, stop, step)

````python
# Example 1: range(stop)

for number in range(5):

    print(number)
````

````python
# Example 2: range(start, stop, step)

for number in range(2,21,2):

    print(number)
````

# 8️⃣ Real-world Examples

````python
# Password Retry System

correct_password = "python123"

for attempt in range(3):

    password = input("Enter Password: ")

    if password == correct_password:
        print("Login Successful")
        break

else:
    print("Account Locked")
````

````python
# Sum of Student Marks

marks = [85,90,78,88,95]

total = 0

for mark in marks:

    total += mark

print("Total Marks =", total)

print("Average =", total/len(marks))
````

# 9️⃣ Pattern Programs

````python
# Square Pattern

for i in range(5):

    for j in range(5):

        print("*", end=" ")

    print()
````

````python
# Right Triangle Pattern

for i in range(1,6):

    for j in range(i):

        print("*", end=" ")

    print()
````
