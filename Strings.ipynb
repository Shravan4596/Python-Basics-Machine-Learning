# 📘 06. Strings in Python

## 📖 Introduction

A **string** is a sequence of characters enclosed in single quotes (`' '`), double quotes (`" "`), or triple quotes (`''' '''` or `""" """`).

Strings are one of the most commonly used data types in Python for handling textual data.

### 💡 Real-world Applications

- User Login Systems
- Text Processing
- Chat Applications
- Email Validation
- Data Cleaning
- Natural Language Processing (NLP)

# 2️⃣ String Indexing

## 🎯 Learning Objective

Access individual characters using their index positions.

Python uses **zero-based indexing**.

````python
text = "Python"

print(text[0])
print(text[2])
print(text[5])
````

````python
# Example 2
word = "Programming"

print(word[-1])
print(word[-2])
print(word[-3])
````

# 3️⃣ String Slicing

## 🎯 Learning Objective

Extract a portion of a string using slicing.

### Syntax

```python
string[start:stop:step]

````python
# Example 1

text = "Python Programming"

print(text[0:6])
````

````python
# Example 2

text = "Python Programming"

print(text[7:])
````

````python
# Example 3

text = "Python"

print(text[::-1])
````

# 4️⃣ String Operators

Python supports several operators for working with strings.

````python
# Example 1: Concatenation

first = "Machine"
second = "Learning"

print(first + " " + second)
````

````python
# Example 2: Repetition

print("Python " * 3)
````

````python
# Example 3: Membership

language = "Python"

print("Py" in language)
print("Java" in language)
````

# 5️⃣ String Methods

Python provides many built-in methods for manipulating strings.

````python
# Example 1: upper()

text = "python"

print(text.upper())
````

````python
# Example 2: lower()

text = "PYTHON"

print(text.lower())
````

````python
# Example 3: title()

text = "machine learning"

print(text.title())
````

````python
# Example 4: capitalize()

text = "python programming"

print(text.capitalize())
````

````python
# Example 5: strip()

text = "   Python   "

print(text.strip())
````

````python
# Example 6: replace()

text = "I like Java"

print(text.replace("Java", "Python"))
````

````python
# Example 7: split()

text = "Python,Java,C++"

print(text.split(","))
````

````python
# Example 8: join()

languages = ["Python", "Java", "C++"]

print(" | ".join(languages))

# Example 9: startswith()

text = "Python Programming"

print(text.startswith("Python"))

# Example 10: endswith()

text = "Python Programming"

print(text.endswith("Programming"))

# Example 11: count()

text = "banana"

print(text.count("a"))

#Example 12: find()

text = "Machine Learning"

print(text.find("Learning"))
````

# 6️⃣ String Formatting

## 🎯 Learning Objective

Learn different ways to format strings in Python.

String formatting allows us to insert variables into strings in a clean and readable way.

Python supports:

- f-strings (Recommended)
- format() method
- % operator (Old Style)

````python
# Example 1: f-String (Recommended)
# Example 1: Using f-string

name = "Shravan"
age = 24

print(f"My name is {name} and I am {age} years old.")


# Example 2: format() Method

language = "Python"
version = 3.12

print("Learning {} Version {}".format(language, version))

# Example 3: Percentage Formatting
# Example 3: Old Style Formatting

name = "Python"

print("Welcome to %s Programming" % name)
````

# 7️⃣ Escape Characters

## 🎯 Learning Objective

Learn how to include special characters inside strings.

Escape characters begin with a backslash (`\`).

````python
# Example 1: New Line and Tab
# Example 1

print("Python\nProgramming")
print("Machine\tLearning")

# Example 2: Quotes Inside String
# Example 2

print("He said \"Python is Awesome\"")
print('It\'s a beautiful day.')

````

# 8️⃣ Real-world String Programs

Let's solve some practical problems using strings.

````python
# Email Validation

email = input("Enter Email: ")

if "@" in email and "." in email:
    print("Valid Email")

else:
    print("Invalid Email")
````

````python
# Palindrome Checker

text = input("Enter a String: ")

if text == text[::-1]:
    print("Palindrome")

else:
    print("Not Palindrome")
````

````python
# Count Vowels

text = input("Enter a String: ")

count = 0

for ch in text.lower():

    if ch in "aeiou":
        count += 1

print("Total Vowels =", count)
````

````python
# Password Strength

password = input("Enter Password: ")

if len(password) >= 8:
    print("Strong Password")

else:
    print("Weak Password")
````

````python
# Count Words

sentence = input("Enter Sentence: ")

words = sentence.split()

print("Total Words =", len(words))
````
