# 1️⃣ Classes and Objects

## 📖 Definition

A **class** is a blueprint or template used to create objects. It defines the attributes (variables) and methods (functions) that the objects created from the class will have.

An **object** is an instance of a class. Each object can have its own data while sharing the behavior defined in the class.

### Why do we use Classes?

- To organize code
- To improve code reusability
- To make programs modular
- To represent real-world entities

### Real-world Examples

- Student
- Employee
- Car
- Bank Account
- Mobile Phone

````python

# Example 1
# Creating a Simple Class

class Student:
    pass

student1 = Student()

print(type(student1))
````

# 2️⃣ Constructors

## 📖 Definition

A **constructor** is a special method in Python that is automatically executed when an object is created.

The constructor is named `__init__()`.

It is mainly used to initialize the object's attributes.

### Syntax

```python
class Student:
    def __init__(self):
        pass
```

### Why do we use Constructors?

- Initialize object data
- Avoid repetitive code
- Automatically assign values during object creation

````python
# Example
class Student:

    def __init__(self, name):
        self.name = name

student = Student("Sravan")

print(student.name)
````

# 3️⃣ Instance Variables

## 📖 Definition

An **instance variable** is a variable that belongs to an individual object.

Each object has its own copy of the instance variables.

Changing the value in one object does not affect another object.

### Real-world Example

Every student has:

- Name
- Roll Number
- Age

These values are different for each student.

````python
# Example
class Student:

    def __init__(self, name):
        self.name = name

s1 = Student("Sravan")
s2 = Student("Rahul")

print(s1.name)
print(s2.name)
````

# 4️⃣ Class Variables

## 📖 Definition

A **class variable** is shared by all objects of the class.

It is defined inside the class but outside any methods.

### Real-world Example

Every student in a college belongs to the same college.

College Name → Class Variable

Student Name → Instance Variable

````python
# Example
class Student:

    college = "ABC College"

    def __init__(self, name):
        self.name = name

s1 = Student("Sravan")
s2 = Student("Rahul")

print(s1.college)
print(s2.college)
````

# 5️⃣ Inheritance

## 📖 Definition

**Inheritance** allows one class (child class) to inherit the properties and methods of another class (parent class).

It promotes code reusability and reduces duplication.

### Advantages

- Code Reusability
- Easy Maintenance
- Extensibility
- Supports Hierarchical Design

### Types of Inheritance

1. Single Inheritance
2. Multiple Inheritance
3. Multilevel Inheritance
4. Hierarchical Inheritance
5. Hybrid Inheritance

````python
# Example 1: Single Inheritance
# Parent Class

class Animal:

    def sound(self):
        print("Animals make sounds.")


# Child Class

class Dog(Animal):

    def bark(self):
        print("Dog barks.")


dog = Dog()

dog.sound()
dog.bark()
````

````python
# Example 2: Multilevel Inheritance
class GrandParent:

    def house(self):
        print("Grandparent's House")


class Parent(GrandParent):

    def car(self):
        print("Parent's Car")


class Child(Parent):

    def bike(self):
        print("Child's Bike")


child = Child()

child.house()
child.car()
child.bike()
````

````python
# Example 3: Multiple Inheritance
class Father:

    def skills(self):
        print("Driving")


class Mother:

    def cooking(self):
        print("Cooking")


class Child(Father, Mother):
    pass


child = Child()

child.skills()
child.cooking()
````

7️⃣ Polymorphism
# 📘 Polymorphism in Python

## 📖 Definition

The word **Polymorphism** means **"Many Forms."**

It allows the same method name to perform different tasks depending on the object that calls it.

Python supports polymorphism through:

- Method Overriding
- Method Overloading (limited support)
- Duck Typing

---

## 🎯 Advantages

- Flexible Code
- Reusable Methods
- Cleaner Programs
- Easy Maintenance

````python
# Example 1: Method Overriding
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog barks")


class Cat(Animal):

    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

````python
# Example 2: Built-in Polymorphism
print(len("Python"))

print(len([10, 20, 30, 40]))

print(len({"A": 1, "B": 2}))
````

````python
# Example 3: Different Objects, Same Method
class Car:

    def move(self):
        print("Car is moving")


class Bike:

    def move(self):
        print("Bike is moving")


for vehicle in (Car(), Bike()):
    vehicle.move()
````

# 📘 Encapsulation in Python

## 📖 Definition

**Encapsulation** is the process of wrapping **data (variables)** and **methods (functions)** into a single unit (class).

It also protects data by restricting direct access using **private variables**.

Private variables start with two underscores (`__`).

---

## 🎯 Advantages

- Data Security
- Data Hiding
- Better Code Organization
- Easy Maintenance

---

## 📝 Syntax

```python
class Student:
    def __init__(self):
        self.__marks = 90
```

````python
# Example 1: Private Variable
class Student:

    def __init__(self):
        self.__marks = 90

    def display(self):
        print("Marks:", self.__marks)


student = Student()

student.display()
````

````python
# Example 2: Getter and Setter Methods
class Employee:

    def __init__(self):
        self.__salary = 0

    def set_salary(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary


employee = Employee()

employee.set_salary(50000)

print(employee.get_salary())
````

````python
# Example 3: Bank Account
class BankAccount:

    def __init__(self):
        self.__balance = 10000

    def deposit(self, amount):
        self.__balance += amount

    def show_balance(self):
        print("Balance:", self.__balance)


account = BankAccount()

account.deposit(5000)

account.show_balance()
````

# 📘 Abstraction in Python

## 📖 Definition

**Abstraction** means **hiding the implementation details** and showing only the essential features to the user.

Python supports abstraction through the **abc (Abstract Base Class)** module.

Abstract classes cannot be instantiated directly. They define methods that must be implemented by child classes.

---

## 🎯 Advantages

- Reduces Complexity
- Improves Security
- Better Code Organization
- Easier Maintenance

---

## 📝 Syntax

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
```

````python
# Example 1: Shape Example
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


rectangle = Rectangle(10, 5)

print("Area =", rectangle.area())
````

````python
# Example 2: Vehicle Example
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts with a key.")


class Bike(Vehicle):

    def start(self):
        print("Bike starts with a self-start button.")


car = Car()
bike = Bike()

car.start()
bike.start()
````

````python
# Example 3: Payment System
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


class UPI(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


credit = CreditCard()
upi = UPI()

credit.pay(2500)
upi.pay(1500)
````
