# 16. Arithmetic Operators

## 🎯 Learning Objective

Understand how to perform basic mathematical operations in Python.

---

## 📖 Theory

Arithmetic operators are used to perform mathematical calculations.

| Operator | Description | Example |
|----------|-------------|---------|
| + | Addition | 10 + 5 |

| * | Multiplication | 10 * 5 |
| / | Division | 10 / 5 |

# 20. Identity Operators

## 🎯 Learning Objective

Understand the difference between equality (`==`) and identity (`is`).

Identity operators compare whether two variables refer to the same object.

````python
# Program 20: Identity Operators

list1 = [10, 20, 30]
list2 = [10, 20, 30]
list3 = list1

print(list1 == list2)
print(list1 is list2)

print(list1 is list3)
print(list1 is not list2)
````
