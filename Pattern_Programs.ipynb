```python
# ⭐ Star Patterns

#Star patterns help you understand nested loops and output formatting.

    # Square Star Pattern

for i in range(5):
    for j in range(5):
        print("*", end=" ")
    print()
```

```python
# Example 2: Right Triangle
# Right Triangle

for i in range(1,6):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Example 3: Inverted Triangle
for i in range(5,0,-1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Example 4: Pyramid
rows = 5

for i in range(rows):
    print(" " * (rows - i - 1), end="")
    print("* " * (i + 1))
```

```python
# Example 5: Inverted Pyramid
rows = 5

for i in range(rows,0,-1):
    print(" " * (rows - i), end="")
    print("* " * i)
```

```python
# 2️⃣ Number Patterns

#Number patterns strengthen your understanding of loop counters and formatting.
# Example 1
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
# Example 2
for i in range(1,6):
    for j in range(i):
        print(i,end=" ")
    print()
```

```python
#Example 3
num = 1

for i in range(1,5):
    for j in range(i):
        print(num,end=" ")
        num += 1
    print()
```

```python
#Example 4
for i in range(5,0,-1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
#Example 5
for i in range(1,6):
    print(str(i) * i)
```

```python
#3️⃣ Alphabet Patterns

# Alphabet patterns combine loops with ASCII characters.
# Example 1
for i in range(65,70):
    for j in range(65,i+1):
        print(chr(j),end=" ")
    print()
```

```python
# Example 2
for i in range(5):
    print(chr(65+i) * (i+1))
```

```python
# Example 3
for i in range(69,64,-1):
    for j in range(65,i+1):
        print(chr(j),end=" ")
    print()
```

```python
#4️⃣ Advanced Patterns
# Example 1: Hollow Square
rows = 5

for i in range(rows):
    for j in range(rows):

        if i == 0 or i == rows-1 or j == 0 or j == rows-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

```python
# Example 2: Hollow Rectangle
rows = 4
cols = 6

for i in range(rows):
    for j in range(cols):

        if i == 0 or i == rows-1 or j == 0 or j == cols-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

```python
#🌍 5️⃣ Real-world Logic Building
#Example 1: Multiplication Table
number = 7

for i in range(1,11):
    print(f"{number} x {i} = {number*i}")
```

```python
#Example 2: Chess Board Pattern
for row in range(8):
    for col in range(8):

        if (row + col) % 2 == 0:
            print("⬜", end=" ")
        else:
            print("⬛", end=" ")

    print()
```

```python
#Example 3: Seating Arrangement
rows = 5
cols = 4

seat = 1

for i in range(rows):
    for j in range(cols):
        print(f"S{seat}", end="\t")
        seat += 1
    print()
```
