# 1️⃣ Lists

## 🎯 Learning Objective

Lists are ordered, mutable (changeable), and allow duplicate values.

### Syntax

```python
list_name = [item1, item2, item3]
```

Lists are the most commonly used collection in Python.

````python
# Example 1: Create a List
# Creating a List

fruits = ["Apple", "Banana", "Mango"]

print(fruits)
````

````python
# Example 2: Access Elements
numbers = [10, 20, 30, 40]

print(numbers[0])
print(numbers[-1])
````

````python
# Example 3: Append Element

languages = ["Python", "Java"]

languages.append("C++")

print(languages)
````

````python
# Example 4: Insert Element

languages = ["Python", "Java"]

languages.insert(1, "SQL")

print(languages)
````

````python
# Example 5: Remove Element

languages = ["Python", "Java", "SQL"]

languages.remove("Java")

print(languages)
````

````python
# Example 6: Pop Element

numbers = [10,20,30,40]

removed = numbers.pop()

print("Removed:", removed)
print(numbers)
````

````python
# Example 7: Sort List

marks = [75,90,60,85]

marks.sort()

print(marks)
````

````python
# Example 8: Reverse List
marks = [10,20,30,40]

marks.reverse()

print(marks)
````

### 💡 Common List Methods

- append()
- insert()
- remove()
- pop()
- sort()
- reverse()
- extend()
- clear()

# 2️⃣ Tuples

## 🎯 Learning Objective

Tuples are ordered but immutable (cannot be modified after creation).

### Syntax

```python
tuple_name = (item1, item2, item3)
```

````python
# Example 1
colors = ("Red","Green","Blue")

print(colors)
````

````python
# Example 2
numbers = (10,20,30,40)

print(numbers[2])
````

````python
# Example 3
numbers = (10,20,30)

print(len(numbers))
````

````python
# Example 4
numbers = (1,2,3,4)

print(max(numbers))
````

````python
# Example 5

numbers = (1,2,3,2,4)

print(numbers.count(2))
````

### 💡 Tuple Characteristics

- Ordered
- Immutable
- Allows duplicates
- Faster than lists

# 3️⃣ Sets

## 🎯 Learning Objective

Sets are unordered collections that store only unique values.

### Syntax

```python
set_name = {item1, item2, item3}
```

````python
# Example 1
numbers = {1,2,3,4}

print(numbers)
````

````python
# Example 2
numbers = {1,2,2,3,3,4}

print(numbers)
````

````python
# Example 3
languages = {"Python","Java"}

languages.add("SQL")

print(languages)
````

````python
# Example 4
languages = {"Python","Java","SQL"}

languages.remove("Java")

print(languages)
````

````python
# Example 5

A = {1,2,3}

B = {3,4,5}

print(A.union(B))
print(A.intersection(B))
````

### 💡 Set Characteristics

- No duplicates
- Unordered
- Mutable
- Fast membership testing

# 4️⃣ Dictionaries

## 🎯 Learning Objective

A dictionary stores data as **key-value pairs**.

### Syntax

```python
dict_name = {
    "key": value
}
```

````python
# Example 1

student = {
    "Name":"Sravan",
    "Age":24,
    "Course":"Python"
}

print(student)
````

````python
# Example 2
student = {
    "Name":"Sravan",
    "Age":24
}

print(student["Name"])
````

````python
# Example 3
student = {"Name":"Sravan"}

student["Age"] = 24

print(student)
````

````python
# Example 4
student = {
    "Name":"Sravan",
    "Age":24
}

student.update({"City":"Hyderabad"})

print(student)
````

````python
# Example 5
student = {
    "Name":"Sravan",
    "Age":24
}

student.pop("Age")

print(student)
````

````python
# Example 6
student = {
    "Name":"Sravan",
    "Age":24
}

print(student.keys())
````

````python
# Example 7
student = {
    "Name":"Sravan",
    "Age":24
}

print(student.values())
````

````python
# Example 8
student = {
    "Name":"Sravan",
    "Age":24
}

for key, value in student.items():
    print(key, ":", value)
````

# 5️⃣ Built-in Functions for Collections

## 🎯 Learning Objective

Python provides several built-in functions that work with Lists, Tuples, Sets, and Dictionaries.

These functions make data processing easier and more efficient.

### Common Built-in Functions

- len()
- max()
- min()
- sum()
- sorted()

````python
# Example 1: len()
# Example 1: Find the Length of a List

languages = ["Python", "Java", "SQL", "C++"]

print("Total Languages:", len(languages))
````

````python
# Example 2: max()
# Example 2: Find Maximum Value

marks = [85, 92, 76, 98, 88]

print("Highest Marks:", max(marks))

# Example 3: min()
# Example 3: Find Minimum Value

marks = [85, 92, 76, 98, 88]

print("Lowest Marks:", min(marks))

# Example 4: sum()
# Example 4: Calculate Total Marks

marks = [85, 92, 76, 98, 88]

print("Total Marks:", sum(marks))

# Example 5: sorted()
# Example 5: Sort a List

numbers = [45, 12, 87, 23, 9]

print("Sorted List:", sorted(numbers))
````

# 6️⃣ Real-world Collection Programs

Collections are widely used in real-world Python applications.

Below are some practical examples.

````python
# Example 1: Student Marks Management
# Student Marks Management

marks = [85, 90, 78, 88, 95]

print("Student Marks:", marks)
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))
print("Average Marks:", sum(marks) / len(marks))
````

````python
# Example 2: Employee Record Using Dictionary
# Employee Information

employee = {
    "ID": 101,
    "Name": "Sravan",
    "Department": "Data Science",
    "Salary": 60000
}

print("Employee Details")

for key, value in employee.items():
    print(f"{key}: {value}")
````

````python
# Example 3: Remove Duplicate Cities
# Remove Duplicate Cities

cities = ["Hyderabad", "Delhi", "Mumbai", "Hyderabad", "Delhi"]

unique_cities = set(cities)

print("Unique Cities:", unique_cities)
````

````python
# Example 4: Shopping Cart
# Shopping Cart

cart = ["Laptop", "Mouse", "Keyboard"]

cart.append("Monitor")

print("Items in Cart")

for item in cart:
    print(item)

print("Total Items:", len(cart))
````

````python
# Example 5: Word Frequency Counter
# Word Frequency Counter

sentence = "python is easy and python is powerful"

words = sentence.split()

frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print("Word Frequency")

for word, count in frequency.items():
    print(f"{word}: {count}")
````
