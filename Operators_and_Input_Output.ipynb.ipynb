# 2. Formatted Output

## 📖 Theory

Python provides multiple ways to format output:

- Comma-separated values
- f-strings (recommended)
- `str.format()`

```python
name = "Shravan"
age = 24

print("Name:", name)
print(f"Age: {age}")
print("Name: {} | Age: {}".format(name, age))
```

# 6. Arithmetic Operators

```python
a = 20
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

# 7. Assignment Operators

```python
x = 10

x += 5
print("After += :", x)

x *= 2
print("After *= :", x)

x -= 10
print("After -= :", x)
```

# 8. Comparison Operators

```python
x = 15
y = 10

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

# 9. Logical Operators

```python
age = 22
salary = 50000

print(age > 18 and salary > 30000)
print(age > 30 or salary > 60000)
print(not age < 18)
```

# 10. Membership Operators

```python
languages = ["Python", "Java", "C++"]

print("Python" in languages)
print("Go" in languages)
print("Go" not in languages)
```

# 11. Identity Operators

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print(list1 == list2)
print(list1 is list2)
print(list1 is list3)
```

# 12. Bitwise Operators

## 📖 Theory

Bitwise operators work on the binary representation of integers.

```python
a = 5
b = 3

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

# Practice Program 1: Calculate the Area of a Rectangle

## 🎯 Learning Objective

Learn how to calculate the area of a rectangle using user input.

---

## 📖 Theory

The area of a rectangle is calculated by multiplying its length and width.

Formula:

Area = Length × Width

---

## 💡 Real-world Example

Calculate the floor area of a room before purchasing tiles or carpet.

```python
# Practice Program 1: Area of Rectangle

length = float(input("Enter the length: "))
width = float(input("Enter the width: "))

area = length * width

print("Area of Rectangle =", area)
```

# Practice Program 2: Convert Celsius to Fahrenheit

## 🎯 Learning Objective

Learn how to convert temperature from Celsius to Fahrenheit.

---

## 📖 Theory

Formula:

Fahrenheit = (Celsius × 9/5) + 32

---

## 💡 Real-world Example

Weather applications often display temperatures in different units.

```python
# Practice Program 2: Celsius to Fahrenheit

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print("Temperature in Fahrenheit =", fahrenheit)
```

# Practice Program 3: Calculate the Average of Three Numbers

## 🎯 Learning Objective

Learn how to calculate the average of multiple values.

---

## 📖 Theory

Formula:

Average = (Number1 + Number2 + Number3) / 3

```python
# Practice Program 3: Average of Three Numbers

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

average = (num1 + num2 + num3) / 3

print("Average =", average)
```

# Practice Program 4: Find the Remainder

## 🎯 Learning Objective

Learn how to use the modulus (`%`) operator.

---

## 📖 Theory

The modulus operator returns the remainder after division.

Example:

10 % 3 = 1

```python
# Practice Program 4: Find the Remainder

dividend = int(input("Enter the dividend: "))
divisor = int(input("Enter the divisor: "))

remainder = dividend % divisor

print("Remainder =", remainder)
```

# Practice Program 5: Check Whether a Number is Even or Odd

## 🎯 Learning Objective

Learn how to determine whether a number is even or odd using the modulus operator.

---

## 📖 Theory

- If a number is divisible by 2, it is **Even**.
- Otherwise, it is **Odd**.

```python
# Practice Program 5: Even or Odd Number

number = int(input("Enter a number: "))

if number % 2 == 0:
    print(number, "is an Even number.")
else:
    print(number, "is an Odd number.")
```
