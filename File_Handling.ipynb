# 1️⃣ Opening Files

Python provides the `open()` function to work with files.

### Syntax

```python
file = open("filename.txt", "mode")
```

The first argument is the file name, and the second argument is the file mode.

````python
# Example 1: Open a File in Read Mode
# Open a file in read mode

file = open("student.txt", "w")
file.write("Hello")
file.close()
````

````python
# Open a file in read mode

file = open("student.txt", "r")

print(file)

file.close()
````

````python
# Example 2: Open a File in Write Mode
# Open a file in write mode

file = open("student.txt", "w")

print("File Opened Successfully")

file.close()
````

````python
#Example 3: Check File Object
file = open("sample.txt", "w")

print(type(file))

file.close()
````

# 2️⃣ Reading Files

Python provides several methods for reading files.

- read()
- readline()
- readlines()

````python
# Example 1: read()
file = open("student.txt", "r")

content = file.read()

print(content)

file.close()
````

````python
# Example 2: readline()
file = open("student.txt", "r")

print(file.readline())

file.close()
````

````python
# Example 3: readlines()
file = open("student.txt", "r")

print(file.readlines())

file.close()
````

````python
# Example 4: Read Using Loop
file = open("student.txt", "r")

for line in file:
    print(line.strip())

file.close()
````

# 3️⃣ Writing Files

The `write()` method writes data to a file.

If the file does not exist, Python creates it

````python
# Example 1
file = open("student.txt", "w")

file.write("Welcome to Python.\n")

file.close()

print("Data Written Successfully")
````

````python
# Example 3
file = open("student.txt", "w")

name = input("Enter Name: ")

file.write(name)

file.close()
````

# 4️⃣ Appending Files

The append mode (`a`) adds new data to the end of an existing file without deleting existing content.

````python
#Example 1
file = open("student.txt", "a")

file.write("\nWelcome Again!")

file.close()
````

````python
# Example 2
name = input("Enter Student Name: ")

file = open("students.txt", "a")

file.write(name + "\n")

file.close()
````

# 5️⃣ File Modes

Python supports different file modes.

Mode	Description
r	Read
w	Write
a	Append
x	Create
r+	Read & Write

````python
# Example 1: Read Mode
file = open("sample.txt", "r")
file.close()
````

````python
# Example 2: Write Mode
file = open("sample.txt", "w")
file.close()
````

````python
# Example 3: Append Mode
file = open("sample.txt", "a")
file.close()
````

````python
#Example 4: Create File
file = open("new_file.txt", "x")
file.close()
````

````python
# Example 5: Read & Write
file = open("sample.txt", "r+")

print(file.read())

file.close()
````

# 6️⃣ Using the with Statement

The `with open()` statement automatically closes the file after use.

It is the recommended way to work with files in Python.

````python
# Example 1
with open("student.txt", "r") as file:
    print(file.read())
````

````python
# Example 2
with open("student.txt", "w") as file:
    file.write("Python Programming")
````

````python
# xample 3
with open("student.txt", "a") as file:
    file.write("\nMachine Learning")
````

# 7️⃣ Working with CSV Files

CSV (Comma-Separated Values) files are commonly used to store tabular data.

````python
# Example 1: Write CSV
import csv

with open("students.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerow(["ID", "Name", "Course"])
    writer.writerow([1, "Sravan", "Python"])
````

````python
# Example 2: Read CSV
import csv

with open("students.csv", "r") as file:

    reader = csv.reader(file)

    for row in reader:
        print(row)
````

````python
# Example 3: Append CSV
import csv

with open("students.csv", "a", newline="") as file:

    writer = csv.writer(file)

    writer.writerow([2, "Rahul", "Machine Learning"])
````

# 8️⃣ Real-world Programs

````python
# Example 1: Student Record System
name = input("Enter Student Name: ")

with open("students.txt", "a") as file:
    file.write(name + "\n")

print("Student Saved Successfully")
````

````python
#Example 2: Employee Details
employee = input("Enter Employee Name: ")

with open("employee.txt", "a") as file:
    file.write(employee + "\n")
````

````python
# Example 3: Count Number of Lines
with open("student.txt", "r") as file:

    lines = file.readlines()

print("Total Lines:", len(lines))
````

````python
# Example 4: Search Word
word = input("Enter Word: ")

with open("student.txt", "r") as file:

    content = file.read()

if word in content:
    print("Word Found")

else:
    print("Word Not Found")
````

````python
# Example 5: Copy File
with open("student.txt", "r") as source:

    content = source.read()

with open("backup.txt", "w") as destination:

    destination.write(content)

print("File Copied Successfully")
````
