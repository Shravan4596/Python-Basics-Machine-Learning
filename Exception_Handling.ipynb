# 📘 Exception Handling in Python
## 📖 Introduction

Exception Handling is a technique used to handle runtime errors without terminating the program.

Instead of crashing, the program handles the error gracefully and continues execution whenever possible.

### Benefits

- Prevents program crashes
- Improves user experience
- Makes code robust
- Easier debugging
- Professional coding practice

# 1️⃣ Common Exceptions

Python provides many built-in exceptions.

Some frequently used exceptions are:

- ZeroDivisionError
- ValueError
- TypeError
- IndexError
- KeyError

```python
# Example 1: ZeroDivisionError
# Division by Zero

try:
    number = 10 / 0
    print(number)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
#Example 2: ValueError
# Invalid Integer Input

try:
    age = int(input("Enter your age: "))
    print(age)

except ValueError:
    print("Please enter a valid integer.")
```

```python
# Example 3: TypeError
# Type Error

try:
    result = "10" + 20

except TypeError:
    print("Cannot add string and integer.")
```

```python
#Example 4: IndexError
numbers = [10, 20, 30]

try:
    print(numbers[5])

except IndexError:
    print("Index is out of range.")
```

```python
#Example 5: KeyError
student = {
    "name": "Sravan",
    "age": 24
}

try:
    print(student["city"])

except KeyError:
    print("Key does not exist.")
```

# 2️⃣ try and except

The `try` block contains code that might raise an exception.

The `except` block handles the exception.

```python
#Example 1
try:
    num = int(input("Enter Number: "))
    print(100 / num)

except Exception as error:
    print("Error:", error)
```

```python
#Example 2
try:
    file = open("sample.txt")

except FileNotFoundError:
    print("File not found.")
```

```python
#Example 3
try:
    numbers = [10, 20]

    print(numbers[3])

except Exception as e:
    print(e)
```

# 3️⃣ else Block

The `else` block executes only if no exception occurs.

```python
#Example 1
try:
    number = int(input("Enter Number: "))

except ValueError:
    print("Invalid Input")

else:
    print("You entered:", number)
```

```python
# Example 2
try:
    result = 100 / 5

except ZeroDivisionError:
    print("Cannot divide")

else:
    print("Result =", result)
```

# 4️⃣ finally Block

The `finally` block always executes, whether an exception occurs or not.

It is commonly used for cleanup operations such as closing files or database connections.

```python
#Example 1
try:
    number = 100 / 10

except Exception:
    print("Error")

finally:
    print("Execution Completed.")
```

```python
# Example 2
try:
    file = open("sample.txt", "r")
    print(file.read())

except FileNotFoundError:
    print("File not found.")

finally:
    print("Closing program.")
```

# 5️⃣ Raising Exceptions

You can raise exceptions manually using the `raise` keyword.

```python
#Example 1
age = int(input("Enter Age: "))

if age < 18:
    raise ValueError("Age must be 18 or above.")

print("Eligible")
```

```python
#Example 2
salary = int(input("Enter Salary: "))

if salary < 0:
    raise ValueError("Salary cannot be negative.")

print("Salary =", salary)
```

# 6️⃣ User-defined Exceptions

Python allows you to create your own exception classes.

```python
#Example 1
class InvalidAgeError(Exception):
    pass

age = int(input("Enter Age: "))

if age < 18:
    raise InvalidAgeError("Not Eligible")

print("Eligible")
```

```python
# Example 2
class InvalidMarks(Exception):
    pass

marks = int(input("Enter Marks: "))

if marks > 100:
    raise InvalidMarks("Marks cannot exceed 100.")

print("Marks =", marks)
```

# 7️⃣ Real-world Examples

```python
#Example 1: ATM Withdrawal
balance = 5000

try:
    amount = int(input("Enter Withdrawal Amount: "))

    if amount > balance:
        raise ValueError("Insufficient Balance")

    print("Transaction Successful")

except ValueError as error:
    print(error)
```

```python
#Example 2: Login Validation
password = input("Enter Password: ")

try:
    if len(password) < 8:
        raise ValueError("Password must contain at least 8 characters.")

    print("Login Successful")

except ValueError as error:
    print(error)
```

```python
# Example 3: Calculator
try:
    a = int(input("First Number: "))
    b = int(input("Second Number: "))

    print("Division =", a / b)

except ZeroDivisionError:
    print("Cannot divide by zero.")

except ValueError:
    print("Invalid input.")
```

```python
# Example 4: File Reading
try:
    with open("student.txt", "r") as file:
        print(file.read())

except FileNotFoundError:
    print("File does not exist.")
```

```python
# Example 5: Dictionary Lookup
student = {
    "Name": "Sravan",
    "Course": "Python"
}

try:
    print(student["Age"])

except KeyError:
    print("Requested key not found.")
```
