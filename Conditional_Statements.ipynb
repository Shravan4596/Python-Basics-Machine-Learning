Python supports the following conditional statements:

- `if`
- `if...else`
- `if...elif...else`
- Nested `if`
- Shorthand `if`

# 1️⃣ The `if` Statement

## 🎯 Learning Objective

Learn how to execute code only when a condition is **True**.

---

## 📖 Theory

The `if` statement is the most basic decision-making statement in Python.

If the condition evaluates to **True**, the code inside the `if` block is executed.

Otherwise, the program skips the block.

### Syntax

```python
if condition:
    statements

````python
# Example 1: Check if a number is positive

number = 15

if number > 14:
    print("The number is positive.")
````

````python
# Example 2: Voting Eligibility

age = 20

if age >= 18:
    print("You are eligible to vote.")
````

````python
# Example 3: Password Length Check

password = "Python123"

if len(password) >= 8:
    print("Password length is valid.")
````

# 2️⃣ The `if...else` Statement

## 🎯 Learning Objective

Learn how to execute one block of code when a condition is **True** and another block when it is **False**.

---

## 📖 Theory

The `if...else` statement allows a program to choose between two possible outcomes.

### Syntax

```python
if condition:
    statements
else:
    statements

````python
# Example 1: Even or Odd Number

number = int(input("Enter a number: "))

if number % 2 == 0:
    print(number, "is an Even number.")
else:
    print(number, "is an Odd number.")
````

````python
# Example 2: Pass or Fail

marks = int(input("Enter your marks: "))

if marks >= 35:
    print("Congratulations! You Passed.")
else:
    print("Sorry! You Failed.")
````

````python
# Example 3: Check Driving Eligibility

age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to apply for a driving license.")
else:
    print("You are not eligible to apply for a driving license.")
````

# 3️⃣ Nested `if` Statement

## 🎯 Learning Objective

Learn how to use one `if` statement inside another `if` statement.

---

## 📖 Theory

A **Nested `if`** is an `if` statement placed inside another `if` statement.

It is useful when multiple conditions must be checked one after another.

### Syntax

```python
if condition1:
    if condition2:
        statements
```

### 💡 Real-world Examples

- ATM PIN Verification
- Login Authentication
- Scholarship Eligibility
- College Admission

````python
# Example 1: ATM Withdrawal Eligibility

balance = 15000
withdraw_amount = 5000

if balance >= withdraw_amount:
    if withdraw_amount <= 10000:
        print("Transaction Successful")
    else:
        print("Daily withdrawal limit exceeded")
else:
    print("Insufficient Balance")
````

````python
# Example 2: Student Scholarship Eligibility

marks = 88
attendance = 92

if marks >= 85:
    if attendance >= 90:
        print("Eligible for Scholarship")
    else:
        print("Attendance is below the required percentage")
else:
    print("Marks are below the scholarship requirement")
````

````python
# Example 3: Login Authentication

username = input("Enter Username: ")
password = input("Enter Password: ")

if username == "admin":
    if password == "admin123":
        print("Login Successful")
    else:
        print("Incorrect Password")
else:
    print("Invalid Username")
````

# 4️⃣ `if...elif...else` Statement

## 🎯 Learning Objective

Learn how to choose one option from multiple conditions.

---

## 📖 Theory

The `if...elif...else` statement is used when there are multiple possible conditions.

Python evaluates conditions from top to bottom.

Once a condition becomes **True**, the remaining conditions are skipped.

### Syntax

```python
if condition:
    statements
elif condition:
    statements
else:
    statements

````python
# Example 1: Grade Calculator

marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 35:
    print("Grade D")
else:
    print("Fail")
````

````python
# Example 2: Age Category

age = int(input("Enter your age: "))

if age < 13:
    print("Child")
elif age < 20:
    print("Teenager")
elif age < 60:
    print("Adult")
else:
    print("Senior Citizen")
````

````python
# Example 3: Largest of Three Numbers

a = int(input("Enter First Number: "))
b = int(input("Enter Second Number: "))
c = int(input("Enter Third Number: "))

if a >= b and a >= c:
    print("Largest Number:", a)
elif b >= a and b >= c:
    print("Largest Number:", b)
else:
    print("Largest Number:", c)
````

#  The `pass` Statement

## 🎯 Learning Objective

Learn how to use the `pass` statement as a placeholder when writing Python programs.

---

## 📖 Theory

Sometimes we know that a block of code will be needed later, but we haven't written it yet.

Python doesn't allow an empty block, so we use the `pass` statement.

The `pass` statement does nothing—it simply prevents a syntax error.

### Syntax

```python
if condition:
    pass
```

### 💡 Real-world Use

- Developing applications step by step
- Creating empty functions
- Creating empty classes
- Placeholder for future implementation

````python
# Example 1: Using pass with if statement

age = 25

if age >= 18:
    pass

print("Program Executed Successfully")
````

#  Multiple Conditions

## 🎯 Learning Objective

Learn how to combine multiple conditions using logical operators.

---

## 📖 Theory

Python provides three logical operators.

| Operator | Meaning |
|-----------|---------|
| and | Both conditions must be True |
| or | At least one condition must be True |
| not | Reverses the result |

These operators are commonly used in login systems, banking applications, and eligibility checks.

````python
# Example 1: Using AND Operator

age = 24
salary = 45000

if age >= 18 and salary >= 30000:
    print("Eligible for Personal Loan")
else:
    print("Not Eligible")
````

````python
# Example 2: Using OR Operator

username = "admin"
password = "python123"

if username == "admin" or password == "admin123":
    print("One condition is True")
else:
    print("Both conditions are False")
````

````python
# Example 3: Using NOT Operator

is_logged_in = False

if not is_logged_in:
    print("Please Login First")
else:
    print("Welcome")
````

## Example 1: ATM Withdrawal System

### 🎯 Objective

Allow withdrawal only if the balance is sufficient and the amount is within the daily limit.

````python
# ATM Withdrawal System

balance = 50000
withdraw = int(input("Enter withdrawal amount: "))

if withdraw <= balance:

    if withdraw <= 10000:

        balance -= withdraw

        print("Transaction Successful")
        print("Remaining Balance:", balance)

    else:
        print("Daily Withdrawal Limit Exceeded")

else:
    print("Insufficient Balance")
````

## Example 4: BMI Category Checker

### 🎯 Objective

Classify a person's BMI into different health categories.

Formula:

BMI = Weight / (Height × Height)

````python
# BMI Calculator

weight = float(input("Enter Weight (kg): "))
height = float(input("Enter Height (meters): "))

bmi = weight / (height ** 2)

print("BMI =", round(bmi, 2))

if bmi < 18.5:
    print("Underweight")

elif bmi < 25:
    print("Normal Weight")

elif bmi < 30:
    print("Overweight")

else:
    print("Obese")
````

## Example 5: Electricity Bill Calculator

### 🎯 Objective

Calculate the electricity bill based on the number of units consumed.

Billing Rules:

- First 100 units → ₹1.50/unit
- Next 100 units → ₹2.50/unit
- Above 200 units → ₹4.00/unit

````python
# Electricity Bill Calculator

units = int(input("Enter Units Consumed: "))

if units <= 100:
    bill = units * 1.5

elif units <= 200:
    bill = (100 * 1.5) + ((units - 100) * 2.5)

else:
    bill = (100 * 1.5) + (100 * 2.5) + ((units - 200) * 4)

print("Electricity Bill = ₹", bill)
````
